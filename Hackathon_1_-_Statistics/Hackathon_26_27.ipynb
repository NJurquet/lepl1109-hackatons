## 2. Fit of distributions

**2.1)** For the 45–64 age group and the winter season only, fit both a Poisson and a Negative Binomial distribution to the variable 'MS\_NUM\_DEATH' using a maximum likelihood approach. Plot histograms overlaid with the fitted probability mass functions. Compare the log-likelihoods and select the most appropriate distribution.

```python
# Code here
# Definition of variables
mort_45_64_winter = mort_clim[(mort_clim["CD_AGEGROUP"] == "45-64") & (mort_clim["season"] == "winter")]
y = mort_45_64_winter["MS_NUM_DEATH"].to_numpy()
n = len(y)
sum_y = np.sum(y)
y_mean = np.mean(y)


# MLE Poisson
def log_likelihood_P(lam, y):
    return np.sum(poisson.logpmf(y, lam))


# MLE Negative Binomial
def log_likelihood_NB(parameters):
    r = np.exp(parameters[0])
    p = expit(parameters[1])
    return -np.sum(nbinom.logpmf(y, r, p))


# Estimator of Poisson
lambda_hat = y_mean
logL_P = log_likelihood_P(lambda_hat, y)

# Estimator of Negative Binomial
var_y = y.var()
if var_y > y_mean:
    r0 = y_mean**2 / (var_y - y_mean)  # Find an initial value for r
else:
    r0 = 1000

p0 = r0 / (r0 + y_mean)

result = minimize(
    log_likelihood_NB, x0=[np.log(r0), np.log(p0 / (1 - p0))], method="L-BFGS-B", bounds=[(-10, 20), (-20, 20)]
)
r_hat = np.exp(result.x[0])  # Because in result we have log(r), so we do exp(x0) = exp(log(r_hat)) -> r_hat = exp(x0)
p_hat = expit(
    result.x[1]
)  # Because in result we have log(p/(1-p)), so we do exp(x1) = exp(log(p_hat/(1-p_hat)) -> p_hat = 1/(1+exp(-x1))

logL_NB = np.sum(nbinom.logpmf(y, r_hat, p_hat))

# Histogram
x = np.arange(y.min(), y.max() + 1)
plt.hist(y, bins=np.arange(y.min() - 0.5, y.max() + 1.5, 1), density=True, alpha=0.5, label="Observed Deaths")

# Poisson PMF
pmf_P = poisson.pmf(x, mu=lambda_hat)
plt.plot(x, pmf_P, color="orange", label="Poisson")

# Negative Binomial PMF
pmf_NB = nbinom.pmf(x, r_hat, p_hat)
plt.plot(x, pmf_NB, color="green", label="Negative Binomial")

plt.xlabel("Number of deaths")
plt.ylabel("Probability")
plt.title("Deaths for age group 45-64 in Winter")
plt.legend()
plt.grid(alpha=0.3)
plt.show()

# Comparison of the log-likelihoods
print("Poisson distribution Log-Likelihood: ", logL_P)
print("Negative Binomial distribution Log-Likelihood: ", logL_NB)

if logL_NB < logL_P:
    print("The log-likelihood of Poisson is preferable")
else:
    print("The log-likelihood of the Negative Binomial is preferable")
```

Comments: 

> With the distributions of Poisson and Negative Binomial, we see that the log-likelihood of the Negative Binomial is higher than that of the Poisson distribution, which means that it is the preferable way to describe this histogram. 
> This is consistent with what we observe on the graph, as it can be seen that the green line is more in line with the data in the center and the extremes.

## 3. Regression

We now aim to assess whether seasonality has an impact on the number of deaths. We consider the following linear model:
$$\log(Y_i) = \beta_0 + \beta_1 X^S_{i1} + \cdots + \beta_3 X^S_{i3} + \beta_4 X^{age}_{i1} + \cdots + \beta_8 X^{age}_{i5}  + \varepsilon_i.$$

where $Y_i$ denotes the number of deaths, $X^S_{ij}$ are dummy variables representing the different seasons, $X^{age}_{ij}$ are dummy variables representing the different age groups, and $\varepsilon_i \sim N(0, \sigma^2)$. 

**3.1)** Why is it more appropriate to use a linear regression on $\log(Y)$ rather than directly on $Y$?

Comments:

> Using log(Y) instead of Y can make the response variable more symmetric and the residuals closer to a normal distribution.
> It can also reduce the influence of extreme values, making the use of linear regression more appropriate.

**3.2)** Perform a linear regression of the number of deaths on the seasonal and group age covariates. Use the `OLS` function from the *statsmodels* package. Make sure to convert the variable *season* and *CD\_AGEGROUP* into dummy variables before fitting the model.

Does seasonality have a significant impact on mortality? Is the mortality similar for each age group? 

```python
# Code here
# Create the variables
y = np.log(mort_clim["MS_NUM_DEATH"]).rename("LOG_DEATHS")  # type: ignore
x = pd.get_dummies(mort_clim[["season", "CD_AGEGROUP"]], drop_first=True, dtype=int)
x = sm.add_constant(x)
print(x.head())  # type: ignore

# OLS regression
model = sm.OLS(y, x).fit()

print(model.summary())

# Season dummy var
season_cols = []
for col in x.columns:  # type: ignore
    if col.startswith("season_"):
        season_cols.append(col)

# Joint F-test: seasonal coef = 0
restrictions = []

for col in season_cols:
    restrictions.append(f"{col} = 0")

restriction = ", ".join(restrictions)

season_test = model.f_test(restriction)

print(season_test)

# Age dummy var
age_cols = []
for col in x.columns:  # type: ignore
    if col.startswith("CD_AGEGROUP_"):
        age_cols.append(col)

# Joint F-test: age coef = 0
restrictions = []

for col in age_cols:
    restrictions.append(f"{col} = 0")

restriction_age = ", ".join(restrictions)
age_test = model.f_test(restriction_age)

print(age_test)
```

Comments: 

> The autumn season and the 0–24 age group were used as reference categories.
>
> Adj. R-squared = 0.951, so this model explains the 95.1% of the variance of the data
>
> _Does seasonality have a significant impact on mortality?_
>
> The joint F-test for the seasonal dummy variables gives (F=17.867) and (p=1.478 * 10^(-11)).
> Since the p < 0.05, we reject the null hypothesis that all seasonal coefficients are zero.
> Therefore, seasonality has a statistically significant association with the number of deaths, controlling for age group.
> In particular, winter is associated with approximately 4.15% more deaths than autumn (found doing %Delta Y = (exp^(0.0407)-1) * 100 ~= 4.25%), while summer is associated with approximately 2.97% fewer deaths (found doing %Delta Y = (exp^(-0.0302) - 1) * 100 ~= -2.97%).
> The spring coefficient is not individually significant at the 5% level.
>
> _Is mortality similar for each age group?_
>
> The joint F-test for the age-group dummy variables gives (F=33755.916), with a p ~= 0.
> We therefore reject the null hypothesis that all age-group coefficients are zero.
> The number of deaths differs significantly across age groups, controlling for season.
> The estimated coefficients increase with age, with the 85+ group showing the largest difference relative to the 0–24 reference group.

# Electricity Consumption and Climate

In this section, we aim to predict electricity consumption using past consumption values and climatic variables. We rely on data provided by ELIA (see https://opendata.elia.be/explore/dataset/ods003/information/?sort=datetime), which records total electricity consumption in Belgium at a 15-minute frequency, and meteorological data from the Royal Meteorological Institute (IRM) (see https://www.meteo.be/fr/services/donnees), which provides daily average temperature observations over Belgium.

The goal is to build a predictive model for electricity demand based on its recent history and weather conditions. From a practical perspective, such forecasting is highly relevant for energy providers and grid operators, as accurately anticipating electricity demand one week ahead helps optimize production planning, ensure grid stability, and reduce operational costs.

## 5. Forecasting 

The goal is to predict electricity consumption at time $t$ using its past values and climate features. Let $Y_t$ denote electricity consumption at time $t$. We consider the following regression model:

$$Y_t = \beta_0 + \beta_1 Y_{t-7Days} + \beta_2 \bar{T}_{t-7Days} + \beta_3 \bar{T}_{t-7Days}^2 + \varepsilon_t,$$

where $Y_{t-7Days}$ denotes electricity consumption observed 7 days earlier, $\bar{T}_{t-7Days}$ is the average temperature observed 7 days earlier $^1$, and $\varepsilon_t \sim \mathcal{N}(0,\sigma^2)$ is an error term.

$^1$ The average temperature is measured at a daily frequency; therefore, $\bar{T}_{t-7Days}$ is constant for all time points $t$ within the same day.

**5.1)** Load the dataset *electricity\_climate.csv* and construct the dataset required to perform the regression.

**Warning:** the data are observed at 15-minute intervals.

To build the regression dataset, each row must contain:
- electricity consumption at time $t$: $Y_t$,
- electricity consumption at time $t - 7$ days: $Y_{t-7Days}$,
- temperature at time $t - 7$ days: $\bar{T}_{t-7Days}$,
- the squared temperature at time $t - 7$ days: $\bar{T}_{t-7Days}^2$.

```python
# Code here
electricity = pd.read_csv(
    "electricity_climate.csv",
    parse_dates=["Datetime"],  # Convert "Datetime" column to datetime format
)
display(electricity.head(n=10))
print("Summary statistics:")
display(electricity.describe(include=[np.number]))  # Display summary statistics without datetime
print(f"Number of NaN/missing values in Consumption: {electricity['Consumption'].isna().sum()}")

time_shift = 7 * 24 * 4  # 7 days * 24 hours * 4 (15 minutes intervals)
elec_dataset = pd.DataFrame(
    {
        "consumption_t": electricity["Consumption"],
        "consumption_t-7days": electricity["Consumption"].shift(time_shift),
        "temp_avg_t-7days": electricity["temp_avg"].shift(time_shift),
        "temp_avg_sq_t-7days": electricity["temp_avg_sq"].shift(time_shift),
    }
).dropna()  # Drop rows with missing Consumption values and those with NaN values resulting from the shift operation as they don't have past data
print("New dataset:")
display(elec_dataset.head(n=10))
```

> The `electricity_climate.csv` dataset is missing 627 electricity consumption values (but the temperature values are complete) accounting for 0.16% of the total number of data rows.
> These rows and their time shifted ones will not allow to give information on the electricity consumption but still provide information on the temperature.
>
> We could either remove these rows from the dataset (loosing information on the temperature and the 15 minutes period) or use an interpolation method to fill the missing values.
>
> As the proportion of missing values is near negligible and the 15 minute frequency not important for the analysis (once the new dataset is created), we will drop/remove rows with missing values from the dataset.
> Temperature is also only used as a supplementary variable to predict electricity consumption, so temperature values are not useful if the corresponding required electricity consumption values are missing.

**5.2)** Split the dataset into a training set (first 75%) and a test set (last 25%).  Use the `OLS` function from the *statsmodels* package to fit linear regression model using only $Y_{t-7Days}$ as explanatory variable.

Analyze the regression output, discuss the statistical significance and performance of the model.

```python
# Code here
# Split the dataset into a training set and a test set.
train_size = int(len(elec_dataset) * 0.75)
elec_dataset_train = elec_dataset.iloc[:train_size]
elec_dataset_test = elec_dataset.iloc[train_size:]

X_train = elec_dataset_train["consumption_t-7days"]  # Explanatory variable
Y_train = elec_dataset_train["consumption_t"]  # Dependent variable
Xm_train = sm.add_constant(X_train)  # Add a constant term for the intercept
elec_model = sm.OLS(Y_train, Xm_train)
elec_results = elec_model.fit()

print(elec_results.summary())

# Plot regression
plt.scatter(X_train, Y_train, alpha=0.5, label="Data", s=10)
plt.plot(X_train, elec_results.predict(Xm_train), color="red", label="Regression")
plt.title("OLS Regression: Electricity Consumption vs. Last Week Consumption")
plt.xlabel("Electricity Consumption at t-7 days [MW]")
plt.ylabel("Electricity Consumption [MW]")
plt.legend()
plt.grid(linestyle="--", alpha=0.5)
plt.show()
```

Comments:

> The regression results provides the value of the regression coefficients $\beta_0$ (intercept) and $\beta_1$ (regression slope).
> The obtained Least Squares regression model is:
> 
> $$ Y_t = \beta_0 + \beta_1 Y_{t-7Days} = 1368.5951 + 0.8342 \ Y_{t-7Days} $$
>
> We can directly sense that there is a non-negligible relation between $Y_t$, the electricity consumption at time t, and $Y_{t-7Days}$, the electricity consumption 7 days earlier.
> The slope $\beta_1$ is positive and near to 1, which indicates that an increase in the electricity consumption at time $t-7$ days should lead to a similar increase (83.42%) in the electricity consumption at time $t$.
>
> The F-statistic is very high (6.492e+05) and therefore, referring to the Fisher distribution, its probability is zero.
> In other words, there is a zero probability on the null hypothesis $H_0$ stating that all variables/coefficients have no effect on the electricity consumption at time t.
> $$ H_0 : \beta_i = 0 \quad \text{ for some } \quad i \in \{1, 2, ..., d\} $$
> As the F-statistic probability is lower than the confidence level $\alpha = 5\%$, we should reject the null hypothesis $H_0$ and conclude that at least one of the coefficients is statistically significant.
> Indeed, the coefficient $\beta_1$ has a p-value of 0, meaning that there is a 0% chance that the electricity consumption 7 days earlier has no effect on the electricity consumption at time t.
> As it is lower than the confidence level $\alpha = 5\%$, $\beta_1$ is significant.
>
> Another important metric is the $R^2$ value, which is a measure of how well the model explains the variability in the dependent variable.
> The closer the $R^2$ value is to 1, the better the model fits the data.
> It is calculated as the ratio of the explained regression variance (SSR) to the total variance (SST).
> $$ R^2 = \frac{SSR}{SST} = 1 - \frac{SSE}{SST} $$
> This regression model has an R-squared value of $R^2 = 0.688$, which means that 68.8% of the variability in electricity consumption at time t can be explained by this model.
> However, 31.2% of the variability still cannot be explained, which suggests that there are other factors influencing electricity consumption that are not understood by this model (such as temperature or measurement errors).
