# Mortality and climate

In this section, we analyze the relationship between mortality, age groups, and seasons in Belgium using publicly available data from Statbel (see https://statbel.fgov.be/en/about-statbel/what-we-do/visualisations/mortality). The dataset contains the number of deaths recorded over time, disaggregated by age group and observed across different seasons. The objective is to study temporal patterns in mortality and assess the role of seasonal effects.

## 1. Basic statistics

**1.1)** Load the dataset mortality\_climate.csv. Convert the variables *season* and *CD\_AGEGROUP* into a categorical variable. 

```python
# Code here
mortality = pd.read_csv("mortality_climate.csv")

display(mortality.head(n=10))  # Display the first 10 rows of the dataset
print(f"Number of data points: {mortality.shape[0]}")
print(f"Number of columns: {mortality.shape[1]}")
print(f"Age groups: {sorted(mortality['CD_AGEGROUP'].unique().tolist())}")
print("\nSummary statistics:")
display(mortality.describe())

# Convert the "date" column to datetime format
mortality = mortality.assign(date=pd.to_datetime(mortality["date"]))

season_categories = sorted(mortality["season"].unique().tolist())
season = pd.Categorical(mortality["season"], categories=season_categories, ordered=True)
age_group_categories = sorted(mortality["CD_AGEGROUP"].unique().tolist())
age_group = pd.Categorical(mortality["CD_AGEGROUP"], categories=age_group_categories, ordered=True)
num_deaths = mortality["MS_NUM_DEATH"]
```

**1.2)** Plot the evolution of the number of deaths (*MS\_NUM\_DEATH*) with respect to age group. What do you observe?

```python
# Code here
# Create a pivot table of number of deaths by date and age group
plot_data = mortality.pivot(index="date", columns="CD_AGEGROUP", values="MS_NUM_DEATH")
display(plot_data)

plot_data.plot(figsize=(12, 6))
plt.title("Evolution of the number of deaths by age group")
plt.xlabel("Date")
plt.ylabel("Number of deaths")
plt.legend(title="Age group")
plt.grid(axis="x", linestyle="--")
plt.show()
```

Comments:

> Firstly, we can directly observe that the general number of deaths increases with age.
> For people aged 0-44, almost no deaths are recorded, while for those aged 85+, the number of deaths is significantly higher.
> This indicates that older people are more vulnerable, and that age is already a strong factor in mortality regardless of the climate/season.
> 
> Then, we can observe a peak in the number of deaths at a regular interval.
> The mortality peaks are happening every year, around the New Year period, which is during the winter season.
> That peak is also more important for older people, which is consistent with the previous observation that older people are more vulnerable.
> This makes sense as winter is a season where people are more likely to get sick.

**1.3)** Do a scatter plot of the number of deaths (*MS\_NUM\_DEATH*) by age group.

```python
# Code here
# Some 0-24 age group data is missing in the dataset.
# Matplotlib is using the order in which labels first appear in the CSV, not the order defined in pd.Categorical.
# To fix 0-24 age group being showed last, we use category `codes` and set ticks labels manually.
plt.scatter(age_group.codes, num_deaths, alpha=0.5)
plt.xticks(
    ticks=range(len(age_group_categories)),
    labels=age_group_categories,
)
plt.title("Number of deaths by age group")
plt.xlabel("Age Group")
plt.ylabel("Number of Deaths")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.show()
```

Comments: 

> This scatter plot shows the distribution of the daily number of deaths across different age groups.
>
> As seen before, the number of deaths for younger age groups (0-24, 25-44) is very low, and older age groups (65-84, 85+) have a much higher number of deaths.
>
> Furthermore, the scatter plot also shows the variability/dispersion in the number of deaths within each age group.
> Younger age groups have a concentrated distribution near zero, while older age groups have a wider spread of values, indicating that the number of deaths can vary significantly from day to day.
> Older age groups may be more sensitive to sudden events such as weather changes or seasonal changes, and disease or virus propagation, which can lead to sudden spikes in mortality.

The goal is to build a predictive model for electricity demand based on its recent history and weather conditions. From a practical perspective, such forecasting is highly relevant for energy providers and grid operators, as accurately anticipating electricity demand one week ahead helps optimize production planning, ensure grid stability, and reduce operational costs.

## 5. Forecasting 

The goal is to predict electricity consumption at time $t$ using its past values and climate features. Let $Y_t$ denote electricity consumption at time $t$. We consider the following regression model:

$$Y_t = \beta_0 + \beta_1 Y_{t-7Days} + \beta_2 \bar{T}_{t-7Days} + \beta_3 \bar{T}_{t-7Days}^2 + \varepsilon_t,$$

where $Y_{t-7Days}$ denotes electricity consumption observed 7 days earlier, $\bar{T}_{t-7Days}$ is the average temperature observed 7 days earlier$^1$, and $\varepsilon_t \sim \mathcal{N}(0,\sigma^2)$ is an error term.

$^1$ The average temperature is measured at a daily frequency; therefore, $\bar{T}_{t-7Days}$ is constant for all time points $t$ within the same day.
